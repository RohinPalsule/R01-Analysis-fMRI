```python
import nibabel as nib
from nilearn import image,datasets,plotting
import glob


def get_diff_z(id):
    z_maps = sorted(glob.glob(f"/Users/rohinpalsule/Desktop/OFC/sub-{id}*"))
    if z_maps is None:
        raise NameError
    late_minus_early = image.math_img(
        "img1 - img2",
        img1=z_maps[1],
        img2=z_maps[0]
    )

    late_minus_early.to_filename(f"results/sub-{id}_late_minus_early_z.nii.gz")

get_diff_z(5169)
```
